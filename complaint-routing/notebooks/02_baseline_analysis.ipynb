```python
import pandas as pd 
```

```python
from sklearn.metrics import confusion_matrix
import joblib

model = joblib.load("../models/baseline_tfidf_lr.joblib")
val = pd.read_parquet("../data/processed/splits/val.parquet")
y_va, X_va = val["product"], val["complaint_what_happened"]

labels = model.classes_
cm = pd.DataFrame(confusion_matrix(y_va, model.predict(X_va), labels=labels),
                  index=labels, columns=labels)
cm.index = cm.index.str.slice(0, 20)
cm.columns = cm.columns.str.slice(0, 12)

print(cm)
```

```python
from collections import Counter

train = pd.read_parquet("../data/processed/splits/train.parquet")
tokens = train["complaint_what_happened"].str.lower().str.findall(r"[a-z']+|\[pii\]")

counts = Counter(t for doc in tokens for t in doc)
total = sum(counts.values())
freqs = sorted(counts.values(), reverse=True)

print("unique words:", len(counts))
for n in [5000, 10000, 20000, 30000, 50000]:
    print(f"top {n:6d}: {sum(freqs[:n]) / total:.2%} of all tokens")
```
